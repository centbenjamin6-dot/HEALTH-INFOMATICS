### Part 6: Targeted Intervention (Identifying High-Risk Individuals) 
#### 6.1 Screening for Cardiac Risks 
In Public Health, we use "Screening Logic" to identify patients who need immediate attention. 
Let's find participants who have High Risk based on three markers: 
1. **Smoker:** Yes
2. **Exercise_Hours_per_Week:** Less than 2 hours 
3. **BMI:** Greater than 30 (Obese range) 


##### Next, we will perform an exploratory data analysis (EDA) focused on the high‑risk group (smoker + low exercise + high BMI) and compare it with the rest of the dataset. This will help understand the profile of the 19 identified individuals.

1. **Load and Prepare Data**
Make sure df is loaded and the high‑risk group is defined as before.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming df is already loaded
high_risk_criteria = (df['Smoker'] == 'Yes') & (df['Exercise_Hours_per_Week'] < 2) & (df['BMI'] > 30)
priority_patients = df[high_risk_criteria].copy()
non_high_risk = df[~high_risk_criteria]

# Extract systolic BP for numeric analysis
priority_patients['Systolic_BP'] = priority_patients['Blood_Pressure'].str.split('/').str[0].astype(float)
non_high_risk['Systolic_BP'] = non_high_risk['Blood_Pressure'].str.split('/').str[0].astype(float)

print(f"High‑risk individuals: {len(priority_patients)}")
print(f"Remaining individuals: {len(non_high_risk)}")
```

2. **Summary Statistics**
Compare means and medians for key numeric variables.

```python
cols = ['Age', 'Heart_Rate', 'Systolic_BP', 'BMI', 'Exercise_Hours_per_Week']
print("=== High‑Risk Group ===")
print(priority_patients[cols].describe())
print("\n=== Non‑High‑Risk Group ===")
print(non_high_risk[cols].describe())
```

**Interpretation:**

1. **Age:** High‑risk group likely older (smoking + high BMI often correlate with age).
2. **Heart Rate:** Possibly elevated due to smoking and low fitness.
3. **Systolic BP:** Expect higher values in high‑risk group.
4. **BMI:** By definition >30 in high‑risk; non‑high‑risk likely lower.
5. **Exercise Hours:** By definition <2 in high‑risk; non‑high‑risk higher.

3. **Visual Comparisons**

Create boxplots to visually compare distributions.

```python
# Create combined dataset for comparison
priority_patients['Group'] = 'High‑Risk'
non_high_risk['Group'] = 'Non‑High‑Risk'

# Extract systolic BP as numeric for both groups (handling possible missing values)
priority_patients['Systolic_BP'] = priority_patients['Blood_Pressure'].str.split('/').str[0]
priority_patients['Systolic_BP'] = pd.to_numeric(priority_patients['Systolic_BP'], errors='coerce')

non_high_risk['Systolic_BP'] = non_high_risk['Blood_Pressure'].str.split('/').str[0]
non_high_risk['Systolic_BP'] = pd.to_numeric(non_high_risk['Systolic_BP'], errors='coerce')

# Combine into one DataFrame for plotting
combined_df = pd.concat([priority_patients, non_high_risk])

# Plot boxplots for key numeric variables with different colors (no warning)
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()
variables = ['Age', 'Heart_Rate', 'Systolic_BP', 'BMI']

for i, var in enumerate(variables):
    sns.boxplot(x='Group', y=var, hue='Group', data=combined_df, 
                palette='Set2', legend=False, ax=axes[i])
    axes[i].set_title(f'{var} by Risk Group')
    axes[i].set_xlabel('Risk Group')
    axes[i].set_ylabel(var)

plt.tight_layout()
plt.show()
```

4. **Categorical Features**

Check diabetic status and gender distribution across groups.

```python
# Add a High_Risk column to the original dataframe
df['High_Risk'] = (df['Smoker'] == 'Yes') & (df['Exercise_Hours_per_Week'] < 2) & (df['BMI'] > 30)

# Now the categorical comparisons work
print("=== Diabetic Status ===")
print(pd.crosstab(df['High_Risk'], df['Diabetic'], normalize='index'))

print("\n=== Gender Distribution ===")
print(pd.crosstab(df['High_Risk'], df['Gender'], normalize='index'))
```

5. **Correlation Within High‑Risk Group**

Understand relationships between risk factors.

```python
corr_matrix = priority_patients[cols].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Correlations within High‑Risk Group')
plt.show()
```

6. **Key Insights**

    a. **Sample size:** Only 19 individuals meet all three criteria – a small but potentially clinically relevant subgroup.

    b. **Age:** High‑risk individuals may be older (check median vs non‑high‑risk).

    c. **Heart rate & BP:** Expect elevated values, indicating cardiovascular strain.

    d. **Diabetic prevalence:** Likely higher among high‑risk due to obesity and smoking.

## **Part 7: Advanced Epidemiological (Correlation)**
## **Understanding the "Web of Causation"**
### **7.1 What is a Correlation Matrix?**
In Public Health, we rarely look at one variable in isolation. We know that Weight affects BMI, and BMI might affect Blood Pressure. A Correlation Matrix calculates a "Correlation Coefficient" ($r$) for every pair of variables: 
1. **+1.0:** Perfect Positive Correlation (As one goes up, the other goes up). 
2. **0.0:** No relationship at all. 
3. **-1.0:** Perfect Negative Correlation (As one goes up, the other goes down—e.g., more Exercise usually means lower Resting Heart Rate). 


```python
import numpy as np

# Step 1: Select only the numerical columns for correlation
# We cannot correlate 'Gender' or 'Smoker' (text) without converting them first 
numeric_df = df.select_dtypes(include=[np.number])

# Step 2: Create the correlation matrix
correlation_matrix = numeric_df.corr()

# Step 3: Display the rounded matrix
correlation_matrix.round(2)
```

### **7.2 Visualizing the Matrix: The Heatmap**
Reading a table of numbers is difficult. 

In Data Science, we use a Heatmap. 

This uses colors to represent the strength of the relationships. 
1. Dark/Hot colors usually indicate strong relationships. 
2. Light/Cool colors indicate weak or negative relationships. 


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the figure size
plt.figure(figsize=(12, 10))

# Create the heatmap using Seaborn
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)

# Add title and display
plt.title('Public Health Heatmap: Variable Inter-correlations', fontsize=16)
plt.show()
```

## **7.3 How to Interpret this Heatmap for Policy**
As a Public Health official, you look for the highest numbers (closest to 1.0 or -1.0). 

1.	**Direct Relationships:** If **Calories_Intake** has a high positive correlation with **Weight_kg**, your policy should focus on nutrition labeling. 

2.	**Protective Factors:** If **Daily_Steps** has a high negative correlation with **Heart_Rate**, you have mathematical proof that walking programs will improve cardiac health in your specific population. 

3.	**Surprises:** Sometimes variables you didn't expect to be related show a strong link, sparking new research questions!
